```python
# ==========================================
# Image Sharpening using High-Boost Filtering
# Google Colab - Single Cell
# ==========================================

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Step 1: Upload Image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Step 2: Read Image
image = cv2.imread(image_path)

# Step 3: Convert to Float (to avoid overflow)
image_float = image.astype(np.float32)

# Step 4: Apply Gaussian Blur
blur = cv2.GaussianBlur(image_float, (5, 5), 0)

# Step 5: Create Mask (Original - Blur)
mask = image_float - blur

# Step 6: High-Boost Filtering
# A > 1 (e.g., 1.5, 2.0)
A = 1.8
high_boost = image_float + A * mask

# Step 7: Convert back to uint8
high_boost = np.clip(high_boost, 0, 255)
high_boost = high_boost.astype(np.uint8)

# Step 8: Show Results
print("Original Image")
cv2_imshow(image)

print("High-Boost Sharpened Image")
cv2_imshow(high_boost)
```
