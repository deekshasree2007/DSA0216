```python
# ==========================================
# Morphological Operation using Morphological Gradient
# Google Colab - Single Cell
# ==========================================

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Step 1: Upload Image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Step 2: Read Image
image = cv2.imread(image_path)

# Step 3: Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Step 4: Convert to Binary Image (Thresholding)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Step 5: Create Structuring Element (Kernel)
kernel = np.ones((5,5), np.uint8)

# Step 6: Apply Morphological Gradient
gradient = cv2.morphologyEx(binary, cv2.MORPH_GRADIENT, kernel)

# Step 7: Display Results
print("Binary Image")
cv2_imshow(binary)

print("Morphological Gradient (Boundary)")
cv2_imshow(gradient)
```
