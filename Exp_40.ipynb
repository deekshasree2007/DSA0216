```python
# ==========================================
# Draw Rectangle and Extract Object using OpenCV
# Google Colab - Single Cell
# ==========================================

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Step 1: Upload Image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Step 2: Read Image
image = cv2.imread(image_path)

# Step 3: Define Rectangle Coordinates (x, y, width, height)
x, y, w, h = 100, 100, 200, 200   # Change values as needed

# Step 4: Draw Rectangle
image_with_rect = image.copy()
cv2.rectangle(image_with_rect, (x, y), (x+w, y+h), (0, 255, 0), 3)

# Step 5: Extract Object (Crop the Region)
extracted_object = image[y:y+h, x:x+w]

# Step 6: Display Results
print("Image with Rectangle")
cv2_imshow(image_with_rect)

print("Extracted Object")
cv2_imshow(extracted_object)
```
