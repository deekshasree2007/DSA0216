```python
# ==========================================
# Insert Watermark to Image using OpenCV
# Google Colab - Single Cell
# ==========================================

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Step 1: Upload Image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Step 2: Read Image
image = cv2.imread(image_path)

# Step 3: Create Overlay for Transparency
overlay = image.copy()

# Step 4: Watermark Text
text = "© My Watermark"
font = cv2.FONT_HERSHEY_SIMPLEX

# Position (bottom-right corner)
(h, w) = image.shape[:2]
text_size = cv2.getTextSize(text, font, 1, 2)[0]
x = w - text_size[0] - 20
y = h - 20

# Add Text to Overlay
cv2.putText(overlay, text, (x, y), font, 1, (255, 255, 255), 2, cv2.LINE_AA)

# Step 5: Blend Original and Overlay
alpha = 0.5   # Transparency level (0 to 1)
watermarked = cv2.addWeighted(overlay, alpha, image, 1 - alpha, 0)

# Step 6: Show Results
print("Original Image")
cv2_imshow(image)

print("Watermarked Image")
cv2_imshow(watermarked)
```
